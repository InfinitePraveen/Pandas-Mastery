```python
import pandas as pd
# 1. Find a messy CSV (or use the UC Berkeley baby names dataset ).
df = pd.read_csv('../data/NationalNames.csv')
df.head(5)
```

```python
# 2. Convert all string columns to title case.
df['Name'] = df['Name'].str.title()
df['Gender'] = df['Gender'].str.title()

# or

# Select all columns with object/string data types and apply title case
string_cols = df.select_dtypes(include=['object', 'string']).columns
df[string_cols] = df[string_cols].astype(str).apply(lambda x: x.str.title())
```

```python
# 3. Create a new column extracting the first 3 letters of each name.
df['Name_Id'] = df['Name'].str.slice(0, 3)
# or
# df['Name'].str[:3]
df
```
